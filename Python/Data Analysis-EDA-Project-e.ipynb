### Number of orders for different months

```python
df['year_month'] = df['invoice_date'].dt.year * 100 + df['invoice_date'].dt.month
df.groupby('invoice_num')['year_month'].first().value_counts().sort_index()
```

### Plot the graph for the above query

```python
import matplotlib.pyplot as plt

orders_month = df.groupby('invoice_num')['year_month'].first().value_counts().sort_index()

ax = orders_month.plot(kind='bar', figsize=(10,3))
ax.set_xlabel('Month and Year', fontsize=10)
ax.set_ylabel('Number of Orders', fontsize=10)
ax.set_title('Number of orders for different Months (1st Dec 2010 - 9th Dec 2011)', fontsize=10)
ax.set_xticklabels(['Dec_10', 'Jan_11', 'Feb_11', 'Mar_11', 'Apr_11', 'May_11', 'Jun_11',
                    'Jul_11', 'Aug_11', 'Sep_11', 'Oct_11', 'Nov_11', 'Dec_11'],
                   rotation='horizontal', fontsize=10)
plt.show()
```

### How many orders (per day)?

```python
df['day'] = df['invoice_date'].dt.dayofweek + 1
df.groupby('invoice_num')['day'].first().value_counts().sort_index()
```

### Day wise sales count

```python
import matplotlib.pyplot as plt

orders_day = df.groupby('invoice_num')['day'].first().value_counts().sort_index()

ax = orders_day.plot(kind='bar', figsize=(7, 3))
ax.set_xlabel('Day', fontsize=8)
ax.set_ylabel('Number of Orders', fontsize=8)
ax.set_title('Number of orders for different Days', fontsize=8)
ax.set_xticklabels(['Mon', 'Tue', 'Wed', 'Thur', 'Fri', 'Sun'], rotation='horizontal', fontsize=8)
plt.show()
```

### How many orders for each country?

```python
group_country_orders = df.groupby('country')['invoice_num'].count().sort_values()
group_country_orders
```

### Draw a horizontal bar graph for above

```python
import matplotlib.pyplot as plt

plt.subplots(figsize=(6, 6))
group_country_orders.plot(kind='barh', fontsize=7)
plt.xlabel('Number of Orders', fontsize=7)
plt.ylabel('Country', fontsize=7)
plt.title('Number of Orders for different Countries', fontsize=7)
plt.show()
```
